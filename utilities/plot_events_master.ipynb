# PLOT EVENTS MASTER

```python
import pandas as pd
import numpy as np
import json
import os
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
os.getcwd()
```

```python
_files = os.listdir('../data/bkup')
```

```python
_files[-1]
```

```python
master_file = f'../data/bkup/{_files[-1]}'
df = pd.read_csv(master_file, encoding = "ISO-8859-1")
df = df.replace(r'^\s+$', np.nan, regex=True) #get rid of unwanted blank spaces in cells...
```

```python
print(df.shape, df.columns)
```

```python
from datetime import date
today = str(date.today())
```

```python
categs = df.columns[8:29]
```

```python
fig, ax = plt.subplots(figsize=(18,11))
df[categs].sum().sort_values(ascending=True).plot(kind='barh', ax=ax)
xcoords = [20,40,60,80]
plt.title(f"Number of Events by Category as of {today}", fontsize=20)
for xc in xcoords:
    plt.axvline(x=xc)
ax.tick_params(axis='both', which='major', labelsize=18)
```
